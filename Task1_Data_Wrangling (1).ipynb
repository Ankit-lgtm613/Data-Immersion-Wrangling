```python

import pandas as pd
import numpy as np

from google.colab import files

uploaded = files.upload()

file_name = next(iter(uploaded))

df = pd.read_excel(file_name)

print("Dataset loaded successfully!")
print("Shape:", df.shape)
display(df.head())
```

```python

# Column names
print("Columns:")
print(df.columns.tolist())

# Data types and non-null values
print("\nDataset information:")
df.info()

# Statistical summary
print("\nStatistical summary:")
display(df.describe(include="all").T)

# Missing values
print("\nMissing values:")
print(df.isnull().sum())

# Exact duplicate rows
print("\nDuplicate rows:", df.duplicated().sum())
```

```python

# Check repeated order IDs
print("Repeated Order_ID rows:")
display(
    df[df["Order_ID"].duplicated(keep=False)]
    .sort_values("Order_ID")
)

# Check categorical values
for col in ["Gender", "City", "Product", "Category"]:
    print(f"\n{col}:")
    print(df[col].value_counts(dropna=False))

# Check invalid age values
print("\nInvalid ages:")
display(df[(df["Age"] < 18) | (df["Age"] > 100)])

# Check sales calculation
difference = (
    df["Total_Sales"] - df["Quantity"] * df["Unit_Price"]
).abs()

print("Incorrect sales calculations:", (difference >= 0.01).sum())
```

```python

# Work on a copy
clean_df = df.copy()

# Standardize text
text_cols = [
    "Order_ID", "Customer_ID", "Customer_Name",
    "Gender", "City", "Product", "Category"
]

for col in text_cols:
    clean_df[col] = clean_df[col].astype("string").str.strip()

for col in ["Gender", "City", "Product", "Category"]:
    clean_df[col] = clean_df[col].str.title()

# Convert date and numeric columns
clean_df["Order_Date"] = pd.to_datetime(
    clean_df["Order_Date"], errors="coerce"
)

for col in ["Age", "Quantity", "Unit_Price", "Total_Sales"]:
    clean_df[col] = pd.to_numeric(
        clean_df[col], errors="coerce"
    )

# Handle missing values
clean_df["Age"] = clean_df["Age"].fillna(
    clean_df["Age"].median()
)

clean_df["City"] = clean_df["City"].fillna(
    clean_df["City"].mode()[0]
)

# Remove exact duplicates only
clean_df = clean_df.drop_duplicates().copy()

# Feature engineering: create age groups
clean_df["Customer_Age_Group"] = pd.cut(
    clean_df["Age"],
    bins=[0, 25, 35, 45, 55, 100],
    labels=["18-25", "26-35", "36-45", "46-55", "56+"]
)

# Validate sales
clean_df["Calculated_Sales"] = (
    clean_df["Quantity"] * clean_df["Unit_Price"]
).round(2)

clean_df["Sales_Matches_Calculation"] = (
    (clean_df["Total_Sales"] - clean_df["Calculated_Sales"]).abs()
    < 0.01
)

print("Cleaning completed!")
display(clean_df.head())
print("Final shape:", clean_df.shape)
print("Missing values remaining:")
print(clean_df.isnull().sum())
```

```python

clean_df.to_csv("cleaned_sales_dataset.csv", index=False)
clean_df.to_excel("cleaned_sales_dataset.xlsx", index=False)

files.download("cleaned_sales_dataset.csv")
```
